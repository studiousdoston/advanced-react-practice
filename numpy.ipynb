```python
import numpy as np

arr =  np.array([1, 3, 6, 8])
print('numpy_array: ', arr)
```

```python
# shape
arr2 =  np.array([[1, 3, 6, 8], [4, 6, 8, 10]])
print('numpy_array2: ', arr2)

arr2.shape
```

```python
# dtype()
a = np.array([1,2,3, 4])
print(a.dtype)
```

```python
# reshape()
a = np.array([1,2,3,4,5,6])
b = a.reshape(3,2)
print('Shaped_array: ', b )
```

```python
# astype()
a = np.array([1,2,3])
b = a.astype(float)
print("float:",  b)
```

```python
# np.zeros() np.ones() -> dummy data
z = np.zeros((4,2))
print('zeros_array:', z)
o = np.ones((3,3))
print('zeros_array:', o)
```

```python
# ndim
a = np.array([[2,3], [4, 6]])
print('ndim: ', a.ndim)
```

```python
# arange()
a = np.arange(1,7)
print('arange:', a)
```

```python
# sum(), mean(), max(), min()

a = np.array([1,2,3,4])
print('sum:', a.sum())
print('mean:', a.mean())
print('max:', a.max())
print('min:', a.min())
```

```python
# flatten()
a = np.array([[[2,3], [3,4], [6,8]]])
print(a.ndim)
print('flattened_3d_array:', a.flatten())
```

```python
# T (Transpose)
a = np.array([[1,2], [3,4]])
print('transpose:', a.T)
```

```python
# dot() @
a = np.array([[1,2], [3,4]])
b = np.array([[2,0], [3,2]])

print("dot/multiplication:", a.dot(b))
```

```python
# random.rand()
a = np.random.rand(2, 3)
print('random rand:', a)
```
